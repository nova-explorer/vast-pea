## Define important parameters
`max_voxel_size=0.1` and `probe_diameter=0.11` were used in the publication to obtain absolute FFV results matching simulation literature.
`max_voxel_size=0.25` and `probe_diameter=1.06` were also used in the publication to obtain absolute FFV results matching experimental literature.

However, decreasing `max_voxel_size` affects the performance of the calculation tremendously.
The runs with `max_voxel_size=0.1` took about 30minutes per data file on a 64 core workstation.
`max_voxel_size=0.25` took about 2minutes per data file on the same workstation.
Both used about 10GB of RAM.

On my laptop, this example with `max_voxel_size=0.25` took about 20minutes.


```python
topofile = '../../data/dilatometry/system_tg.600.data' # LAMMPS data file to be read
max_voxel_size = 1.0 # maximum size of the each voxel used to calculate FFV
probe_diameter = 1.1 # size of the probe inserted in each voxel to determine if they are free or not
```
