# Example calculation of crosslink density

## Crosslink density for DGEBA/AFD using bounds

```python
# Read LAMMPS data file
lammps_data = data_file(path="../../data/preparation/", pattern="afd_annealing2.data")

# Compute crosslink density for this data file
compute_crosslink_density(lammps_data,
                             effective=True, # whether network strands are counted as elastically effective or not
                             bounds=lammps_data.get_bounds(), # provide bounds to calculate volume
                             crosslink_point_type=10, # type of the crosslink points. Here this corresponds to the nn type
                             crosslink_types=[3] # types of the crosslinks connected to the crosslink point. Here this corresponds to the c3 type. The phenyl-disulfide-phenyl crosslinks are implied.
                             )
```

## Crosslink density for DGEBA/AFD using volume

```python
# See explanations for this in 0_preprocess
from lammps_traj import trajectory
traj = trajectory(path="../../data/BER/",
                  pattern="afd_ber_325K.dump",
                  filter_type=False,
                  filter=False)

# Get an average volume value by averaging bounds over timesteps and multiplying them
# Could also be possible to get that from a thermo log
volume = float(traj.get_bounds().mean(dim="ts").prod())

compute_crosslink_density(lammps_data,
                             effective=True,
                             volume=volume,
                             crosslink_point_type=10,
                             crosslink_types=[3]
                             )
```

## Crosslink density for a hypothetical 100% conversion

```python
# Compute crosslink density for a hypothetical 100% conversion system
from constants import AVOGADRO_NUMBER, ANGSTROM_TO_METER

n_hardener = 376
eff_strands_per_hardener = 3

n_hardener * eff_strands_per_hardener / volume / ANGSTROM_TO_METER**3 / AVOGADRO_NUMBER
```
